# Data

```python
import pandas as pd
import numpy as np

file_paths = ["Data/Altitude50","Data/Altitude100", "Data/Altitude150"]

columns = ["t [days]", "a [km]", "ecc", "i [deg]", "RAAN [deg]", "argp [deg]","Altitude [km]", "nu [deg]"]
df50 = pd.read_csv(file_paths[0], names=columns, skiprows=1)
df100 = pd.read_csv(file_paths[1], names=columns, skiprows=1)
df150 = pd.read_csv(file_paths[2], names=columns, skiprows=1)
```

# Plots

```python
import matplotlib.pyplot as plt

curve_colors = ['tab:blue','tab:orange', 'tab:green']
plot_style = {
    'grid_color': 'gray',
    'grid_alpha': 0.5,
    'text_color': 'white'
}
```

## Altitude

```python
plt.figure(figsize=(10, 3), facecolor='black')
plt.plot(df50['t [days]'], df50['Altitude [km]'], color=curve_colors[0], label='50 km', linewidth=2,zorder=3, alpha=0.8)
plt.plot(df100['t [days]'], df100['Altitude [km]'], color=curve_colors[1], label='100 km', linewidth=2, zorder=2, alpha=0.9)
plt.plot(df150['t [days]'], df150['Altitude [km]'], color=curve_colors[2], label='150 km', linewidth=2, zorder=1, alpha=1.0)
plt.grid(color=plot_style['grid_color'], alpha=plot_style['grid_alpha'])
plt.xlabel('Tid [dager]', color=plot_style['text_color'], fontsize=14)
plt.ylabel('Høyde [km]', color=plot_style['text_color'], fontsize=14)
plt.ylim(0, max(df150['Altitude [km]']) + 10)
plt.xticks(np.arange(0, max(df150['t [days]']), 10), color=plot_style['text_color'])
plt.xlim(0, max(df150['t [days]']))
plt.legend(loc='best', fontsize=8, facecolor='black', edgecolor='white', framealpha=0.8, shadow=True, title='Initiell Høyde', title_fontsize=10)
plt.savefig('Figurer/tidsserie_alt.pdf', bbox_inches='tight')
plt.show()
```

## Delta Altitude

```python
plt.figure(figsize=(10, 3), facecolor='black')
plt.plot(df50['t [days]'], df50['Altitude [km]']-50, color=curve_colors[0], label='50 km', linewidth=2,zorder=1, alpha=1)
plt.plot(df100['t [days]'], df100['Altitude [km]']-100, color=curve_colors[1], label='100 km', linewidth=2, zorder=2, alpha=0.8)
plt.plot(df150['t [days]'], df150['Altitude [km]']-150, color=curve_colors[2], label='150 km', linewidth=2, zorder=3, alpha=0.6)
plt.grid(color=plot_style['grid_color'], alpha=plot_style['grid_alpha'])
plt.xlabel('Tid [dager]', color=plot_style['text_color'], fontsize=14)
plt.ylabel(r'$\Delta$ Høyde [km]', color=plot_style['text_color'], fontsize=14)
plt.ylim(-150,150 + 10)
plt.xticks(np.arange(0, max(df150['t [days]']), 10), color=plot_style['text_color'])
plt.xlim(0, max(df150['t [days]']))
plt.legend(loc='best', fontsize=8, facecolor='black', edgecolor='white', framealpha=0.8, shadow=True, title='Initiell Høyde', title_fontsize=10)
plt.savefig('Figurer/tidsserie_delta_alt.pdf', bbox_inches='tight')
plt.show()
```

## Argument of Perigee

```python
df50['argp_unwrapped'] = np.rad2deg(np.unwrap(np.deg2rad(df50['argp [deg]']))) 
df100['argp_unwrapped'] = np.rad2deg(np.unwrap(np.deg2rad(df100['argp [deg]']))) 
df150['argp_unwrapped'] = np.rad2deg(np.unwrap(np.deg2rad(df150['argp [deg]']))) 

plt.figure(figsize=(10, 3), facecolor='black')
plt.plot(df50['t [days]'], df50['argp_unwrapped'], color=curve_colors[0], label='50 km', linewidth=2,zorder=3, alpha=0.8)
plt.plot(df100['t [days]'], df100['argp_unwrapped'], color=curve_colors[1], label='100 km', linewidth=2, zorder=2, alpha=0.9)
plt.plot(df150['t [days]'], df150['argp_unwrapped'], color=curve_colors[2], label='150 km', linewidth=2, zorder=1, alpha=1.0)

plt.grid(color=plot_style['grid_color'], alpha=plot_style['grid_alpha'])
plt.xlabel('Tid [dager]', color=plot_style['text_color'], fontsize=14)
plt.ylabel(r'Argp ($\omega$) [deg]', color=plot_style['text_color'], fontsize=14)
plt.xticks(np.arange(0, max(df150['t [days]']), 10), color=plot_style['text_color'])
plt.xlim(0, max(df150['t [days]']))
plt.legend(loc='best', fontsize=8, facecolor='black', edgecolor='white', framealpha=0.8, shadow=True, title='Initiell Høyde', title_fontsize=10)

plt.savefig('Figurer/tidsserie_argp.pdf', bbox_inches='tight')
plt.show()
```
